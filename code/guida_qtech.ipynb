## Guida Rapida per QTech

#### ***Collegarsi a QTech***

Per connettersi al server QTech dell'università tramite SSH dal terminale:
```bash 
ssh username@qtech.fisica.unimi.it
```

#### ***Scaricare per la prima volta la Cartella Progetto su QTech***

Se non hai ancora la tua repository di GitHub sul server QTech, devi clonarla nella tua cartella home. Entra nel server e lancia:
```shell
git clone git@github.com:tuo-username/nome-repo.git
```
Il codice git@github si trova nella sezione Code di github

#### ***Importare modifiche da PC a QTech tramite Github***

Ogni volta che modifichi i file su PC devi caricare i dati su Github, facendo in questo ordine:
```shell
git add .
git commit -m "Descrizione Aggiornamento"
git push
```
A questo punto puoi collegarti su QTech con ssh, entra nella cartella principale del progetto e fai:
```shell
git pull origin main
```
Per controllare di aver importato tutte le modifiche fai:
```shell
git status
```
Allo stesso modo, se vuoi spostare i dati dal terminale QTech a quello personale, fai questo identico procedimento scambiando i ruoli.

#### ***Inviare file DA QTech a PC***

Se sei connesso in SSH dentro QTech e vuoi importare una cartella direttamente sul PC, usa:

```shell
scp -r username@qtech.fisica.unimi.it:"/percorso_cartella" ~/percorso_destinazione
```


#### ***Scaricare Julia su QTech***

Se Julia non è presente sul server QTech:
```shell
curl -fsSL https://install.julialang.org | sh
```
Per controllare se è stato installato:
```shell
julia
```
A questo punto nel terminale Julia è possibile gestire i pacchetti. Se hai un file <code>Project.toml</code> nella cartella, attiva l'ambiente locale e scarica i pacchetti corretti digitando:

```shell
pkg> activate .
pkg> instantiate
```
Oppure aggiungi un pacchetto necessario con
```shell
pkg> add ProgressMeter
```

#### ***Far Partire i programmi su QTech***

Per far partire un programma con QTech bisogna scrivere il file di sottomissione quindi fai:
```shell
nano submit.sh
```
All'interno di questo file andranno scritte le istruzioni per la sottomissione, copia e incolla questo:
```shell
#!/bin/bash
#SBATCH --job-name=calcolo_julia
#SBATCH --output=risultato_%j.log
#SBATCH --error=errore_%j.log
#SBATCH --ntasks=1
#SBATCH --cpus-per-task=1
#SBATCH --mem=2G

julia --project=. nome_script.jl
```
Dove
- <code>job-name</code>: Nome del job che vedrai in squeue
- <code>output</code>: File dove finirà il testo del println
- <code>error</code>: File dove finiranno i log e le barre di caricamento
- <code>ntasks</code>: Indica il numero di istanze indipendenti del programma da lanciare, anche questo si setta tendenzialmente a 1 tranne in caso di calcolo parallelo.
- <code>cpus-per-task</code>: Indica quanti core della macchina devono essere riservati per il tuo singolo task (meglio 1 a meno che tu faccia calcolo parallelo)
- <code>mem</code>: Limite massimo di memoria RAM che il server bloccherà per il tuo calcolo. Più ne metti più vieni messo dopo in coda, se lo superi il programma termine.

#### ***Lanciare il programma***
Dopo aver salvato il file submit.sh fai:
```shell
sbatch submit.sh
```

#### ***Monitorare il programma***
A questo punto puoi controllare lo stato della coda con:
```shell
queue
```
Se il tuo programma sta andando vedrai un indicatore "R" e da quanto tempo sta girando. Se la coda è vuota ha finito o c'è stato un errore, controlla dai file di log che si creano automaticamente nella cartella di esecuzione.
